# 05C — S3E + Cosine-Utility MBR — Fine-Tuning LoRA (Locked) — FINAL V1

Notebook ini memproses **265 kandidat final (53 × 5)** dari skenario **Fine-Tuning LoRA (Locked)**.

Notebook ini:
- tidak menjalankan Llama lagi;
- tidak membaca reference Bahasa Indonesia;
- hanya memvalidasi kandidat, membentuk embedding MPNet, menghitung cosine similarity, S3E, dan MBR;
- belum menghitung BLEU/chrF++.

**S3E**  
`H = -(1/n) Σ_i log[(1/(n-1)) Σ_(j≠i) S(y_i,y_j)]`

**MBR utility**  
`U(y_i) = (1/(n-1)) Σ_(j≠i) S(y_i,y_j)`

Ketentuan:
- model embedding: `sentence-transformers/paraphrase-multilingual-mpnet-base-v2`;
- embedding L2-normalized, dimensi 768;
- cosine negatif → 0;
- diagonal disimpan 1 tetapi tidak dipakai dalam mean;
- `epsilon=1e-12`;
- tie MBR → candidate index terkecil;
- S3E berjalan paralel dengan MBR dan tidak menjadi input seleksi.

Output:
`Hasil/final_selection/lora_locked/<run_input>/`

## 3. Temukan + audit kandidat final

```python
FORBIDDEN_REFERENCE_COLUMNS = {
    "reference",
    "reference_text",
    "indonesia_text",
    "gold",
    "gold_reference",
}

def read_json_if_exists(path: Path):
    if not path.exists():
        return None
    try:
        return json.loads(
            path.read_text(encoding="utf-8")
        )
    except Exception:
        return None


def canonicalize(path: Path) -> pd.DataFrame:
    raw = pd.read_csv(
        path,
        dtype={"row_id": str},
        keep_default_na=False,
        encoding="utf-8-sig",
    )

    leaked = (
        FORBIDDEN_REFERENCE_COLUMNS
        & set(raw.columns)
    )
    if leaked:
        raise RuntimeError(
            f"Reference column ditemukan: {sorted(leaked)}"
        )

    if "lora" == "classic":
        required = {
            "scenario",
            "row_id",
            "candidate_id",
            "seed",
            "source",
            "output",
        }
        missing = required - set(raw.columns)
        if missing:
            raise RuntimeError(
                f"Kolom kurang: {sorted(missing)}"
            )

        df = pd.DataFrame({
            "scenario": raw["scenario"].astype(str),
            "row_id": raw["row_id"].astype(str),
            "conversation_no":
                raw["conversation_no"].astype(str)
                if "conversation_no" in raw.columns else "",
            "turn_order":
                raw["turn_order"].astype(str)
                if "turn_order" in raw.columns else "",
            "source_text": raw["source"].astype(str),
            "candidate_index":
                pd.to_numeric(
                    raw["candidate_id"],
                    errors="raise",
                ).astype(int),
            "seed":
                pd.to_numeric(
                    raw["seed"],
                    errors="raise",
                ).astype(int),
            "candidate_text": raw["output"].astype(str),
            "technical_error":
                raw["error"].astype(str)
                if "error" in raw.columns else "",
        })

        for col in [
            "temperature",
            "top_p",
            "top_k",
            "max_new_tokens",
            "coverage",
            "fragment_count",
            "example_count",
            "max_fragment_len",
            "k_per_fragment",
            "prompt_tokens",
            "completion_tokens",
            "latency_seconds",
        ]:
            if col in raw.columns:
                df[col] = raw[col].values

    else:
        required = {
            "scenario",
            "row_id",
            "candidate_index",
            "seed",
            "duri_text",
            "candidate_text",
        }
        missing = required - set(raw.columns)
        if missing:
            raise RuntimeError(
                f"Kolom kurang: {sorted(missing)}"
            )

        df = pd.DataFrame({
            "scenario": raw["scenario"].astype(str),
            "row_id": raw["row_id"].astype(str),
            "conversation_no":
                raw["conversation_no"].astype(str)
                if "conversation_no" in raw.columns else "",
            "turn_order":
                raw["turn_order"].astype(str)
                if "turn_order" in raw.columns else "",
            "source_text": raw["duri_text"].astype(str),
            "candidate_index":
                pd.to_numeric(
                    raw["candidate_index"],
                    errors="raise",
                ).astype(int),
            "seed":
                pd.to_numeric(
                    raw["seed"],
                    errors="raise",
                ).astype(int),
            "candidate_text":
                raw["candidate_text"].astype(str),
            "technical_error": "",
        })

        for col in [
            "temperature",
            "top_p",
            "top_k",
            "max_new_tokens",
            "generation_time_seconds",
        ]:
            if col in raw.columns:
                df[col] = raw[col].values

        if "is_empty" in raw.columns:
            empty_mask = (
                raw["is_empty"]
                .astype(str)
                .str.casefold()
                .isin({"true", "1", "yes"})
            )
            df.loc[
                empty_mask,
                "technical_error",
            ] = "empty_output"

    return df


def audit_candidates(df: pd.DataFrame):
    if len(df) != EXPECTED_TOTAL_CANDIDATES:
        raise RuntimeError(
            f"Jumlah kandidat {len(df)} != 265."
        )

    if df["row_id"].nunique() != EXPECTED_TEST_ROWS:
        raise RuntimeError(
            f"Unique row_id {df['row_id'].nunique()} != 53."
        )

    counts = df.groupby("row_id").size()
    if not (counts == 5).all():
        raise RuntimeError(
            "Tidak semua row_id memiliki 5 kandidat."
        )

    if set(df["seed"].astype(int).unique()) != EXPECTED_SEEDS:
        raise RuntimeError(
            f"Seed tidak sesuai: "
            f"{sorted(df['seed'].astype(int).unique())}"
        )

    if df[
        ["row_id", "candidate_index", "seed"]
    ].duplicated().any():
        raise RuntimeError(
            "Duplikasi row_id/candidate_index/seed."
        )

    for row_id, group in df.groupby(
        "row_id",
        sort=False,
    ):
        if set(
            group["candidate_index"].astype(int)
        ) != {1, 2, 3, 4, 5}:
            raise RuntimeError(
                f"{row_id}: candidate_index bukan 1..5."
            )

    if "temperature" in df.columns:
        vals = pd.to_numeric(
            df["temperature"],
            errors="coerce",
        ).dropna()
        if len(vals) and not np.allclose(
            vals.astype(float),
            0.70,
            atol=1e-12,
            rtol=0,
        ):
            raise RuntimeError(
                "temperature bukan 0.70."
            )

    if "top_p" in df.columns:
        vals = pd.to_numeric(
            df["top_p"],
            errors="coerce",
        ).dropna()
        if len(vals) and not np.allclose(
            vals.astype(float),
            0.95,
            atol=1e-12,
            rtol=0,
        ):
            raise RuntimeError(
                "top_p bukan 0.95."
            )

    if "top_k" in df.columns:
        vals = pd.to_numeric(
            df["top_k"],
            errors="coerce",
        ).dropna()
        if len(vals) and not np.allclose(
            vals.astype(float),
            0.0,
            atol=1e-12,
            rtol=0,
        ):
            raise RuntimeError(
                "top_k bukan 0."
            )

    if "max_new_tokens" in df.columns:
        vals = pd.to_numeric(
            df["max_new_tokens"],
            errors="coerce",
        ).dropna()
        if len(vals) and not np.allclose(
            vals.astype(float),
            128.0,
            atol=1e-12,
            rtol=0,
        ):
            raise RuntimeError(
                "max_new_tokens bukan 128."
            )


candidate_paths = sorted(
    INPUT_SCENARIO_DIR.glob(
        "run_*/candidates_long.csv"
    ),
    key=lambda p: p.stat().st_mtime,
    reverse=True,
)

valid_inputs = []

for path in candidate_paths:
    try:
        df_try = canonicalize(path)
        audit_candidates(df_try)

        scenario_values = set(
            df_try["scenario"].astype(str).unique()
        )
        if scenario_values != {"lora_locked"}:
            raise RuntimeError(
                f"Scenario tidak sesuai: {scenario_values}"
            )

        run_dir = path.parent
        manifest = (
            read_json_if_exists(
                run_dir / "run_manifest.json"
            )
            or read_json_if_exists(
                run_dir / "manifest.json"
            )
        )

        if manifest is None:
            raise RuntimeError(
                "Manifest generation tidak ditemukan."
            )

        ref_flag = manifest.get(
            "reference_read",
            manifest.get(
                "test_reference_read",
                None,
            ),
        )
        if ref_flag is not False:
            raise RuntimeError(
                "Manifest tidak membuktikan reference_read=False."
            )

        seeds_manifest = manifest.get(
            "candidate_seeds",
            manifest.get("seeds", []),
        )
        if set(map(int, seeds_manifest)) != EXPECTED_SEEDS:
            raise RuntimeError(
                f"Seed manifest tidak sesuai: {seeds_manifest}"
            )

        gen = manifest.get("generation", {})
        if not np.isclose(
            float(gen.get("temperature", -1)),
            0.70,
        ):
            raise RuntimeError(
                "Manifest temperature bukan 0.70."
            )
        if not np.isclose(
            float(gen.get("top_p", -1)),
            0.95,
        ):
            raise RuntimeError(
                "Manifest top_p bukan 0.95."
            )
        if int(gen.get("top_k", -999)) != 0:
            raise RuntimeError(
                "Manifest top_k bukan 0."
            )
        if int(
            gen.get("max_new_tokens", -1)
        ) != 128:
            raise RuntimeError(
                "Manifest max_new_tokens bukan 128."
            )
        if (
            "num_beams" in gen
            and int(gen["num_beams"]) != 1
        ):
            raise RuntimeError(
                "Manifest num_beams bukan 1."
            )

        valid_inputs.append(
            (path, df_try, manifest)
        )

    except Exception as exc:
        print(
            "Skip:",
            path,
            "|",
            type(exc).__name__,
            str(exc)[:220],
        )

if not valid_inputs:
    raise FileNotFoundError(
        "Tidak ditemukan final candidates_long.csv "
        "yang lolos audit."
    )

CANDIDATES_PATH, candidates, generation_manifest = (
    valid_inputs[0]
)
INPUT_RUN_DIR = CANDIDATES_PATH.parent
INPUT_RUN_NAME = INPUT_RUN_DIR.name
OUTPUT_DIR = OUTPUT_ROOT / INPUT_RUN_NAME
OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

print("\nINPUT FINAL TERPILIH")
print("Candidates :", CANDIDATES_PATH)
print("Run        :", INPUT_RUN_NAME)
print("Rows       :", len(candidates))
print("Unique test:", candidates["row_id"].nunique())
print("Reference  : TIDAK ADA / TIDAK DIBACA")
print("Output dir :", OUTPUT_DIR)

display(candidates.head(10))
```

## 4. Validasi kandidat

```python
REFUSAL_PATTERNS = [
    r"^maaf\b",
    r"^mohon maaf\b",
    r"\bsaya tidak (mengerti|memahami|dapat|bisa)\b",
    r"\btidak dapat menerjemahkan\b",
    r"\btidak bisa menerjemahkan\b",
    r"\bi cannot\b",
    r"\bi can't\b",
]

def candidate_validity(
    text: str,
    technical_error: str,
):
    if str(technical_error).strip():
        return "invalid", "technical_failure"

    normalized = re.sub(
        r"\s+",
        " ",
        str(text).strip(),
    )

    if len(normalized) < 2:
        return "invalid", "empty_or_too_short"

    low = normalized.casefold()

    for pattern in REFUSAL_PATTERNS:
        if re.search(
            pattern,
            low,
            flags=re.IGNORECASE,
        ):
            return "invalid", "refusal"

    return "valid", ""


results = candidates.apply(
    lambda r: candidate_validity(
        r["candidate_text"],
        r["technical_error"],
    ),
    axis=1,
)

candidates["semantic_status"] = [
    x[0] for x in results
]
candidates["semantic_reason"] = [
    x[1] for x in results
]

print(
    candidates[
        ["semantic_status", "semantic_reason"]
    ].value_counts(dropna=False)
)

invalid_preview = candidates[
    candidates["semantic_status"] != "valid"
].copy()

if not invalid_preview.empty:
    display(
        invalid_preview[
            [
                "row_id",
                "candidate_index",
                "seed",
                "candidate_text",
                "semantic_reason",
            ]
        ].head(30)
    )
else:
    print("Semua 265 kandidat valid.")
```

## 5. Load MPNet

```python
print(
    "Load embedding model:",
    EMBEDDING_MODEL_ID,
)

embedder = SentenceTransformer(
    EMBEDDING_MODEL_ID,
    device=EMBEDDING_DEVICE,
    cache_folder=str(EMBEDDING_CACHE),
    trust_remote_code=False,
)

probe = embedder.encode(
    ["Ini kalimat uji."],
    convert_to_numpy=True,
    normalize_embeddings=True,
    show_progress_bar=False,
)

if probe.shape != (1, 768):
    raise RuntimeError(
        f"Dimensi embedding tidak sesuai: {probe.shape}"
    )

probe_norm = float(
    np.linalg.norm(probe[0])
)

if not np.isclose(
    probe_norm,
    1.0,
    atol=1e-4,
):
    raise RuntimeError(
        f"Embedding belum L2-normalized: {probe_norm}"
    )

print("Embedding dim : 768")
print("Probe norm    :", probe_norm)
print("EMBEDDING MODEL: READY")
```

## 6. Hitung similarity, S3E, dan MBR

```python
detail_rows = []
selected_rows = []
pair_rows = []
matrix_rows = []

row_order = (
    candidates[["row_id"]]
    .drop_duplicates()["row_id"]
    .tolist()
)

for row_id in row_order:
    group = (
        candidates[
            candidates["row_id"] == row_id
        ]
        .copy()
        .sort_values(
            "candidate_index",
            kind="stable",
        )
    )

    valid = (
        group[
            group["semantic_status"] == "valid"
        ]
        .copy()
        .sort_values(
            "candidate_index",
            kind="stable",
        )
    )

    n = len(valid)
    utilities = {}
    s3e_value = np.nan
    selected_candidate_index = np.nan
    selected_output = ""
    selected_utility = np.nan

    if n >= 2:
        texts = valid[
            "candidate_text"
        ].astype(str).tolist()

        E = embedder.encode(
            texts,
            batch_size=EMBEDDING_BATCH_SIZE,
            convert_to_numpy=True,
            normalize_embeddings=True,
            show_progress_bar=False,
        ).astype(np.float64)

        norms = np.linalg.norm(E, axis=1)
        if not np.allclose(
            norms,
            1.0,
            atol=1e-4,
        ):
            raise RuntimeError(
                f"{row_id}: embedding tidak normalized."
            )

        sim = np.clip(
            E @ E.T,
            0.0,
            1.0,
        )
        np.fill_diagonal(sim, 1.0)

        sim_off = sim.copy()
        np.fill_diagonal(sim_off, 0.0)

        mean_off = (
            sim_off.sum(axis=1)
            / (n - 1)
        )

        s3e_value = float(
            -np.mean(
                np.log(
                    np.maximum(
                        mean_off,
                        EPS,
                    )
                )
            )
        )

        valid_indices = (
            valid["candidate_index"]
            .astype(int)
            .tolist()
        )

        utilities = {
            int(valid_indices[i]):
                float(mean_off[i])
            for i in range(n)
        }

        best_utility = max(
            utilities.values()
        )
        tied = [
            idx
            for idx, utility
            in utilities.items()
            if abs(
                utility - best_utility
            ) <= TIE_TOL
        ]
        selected_candidate_index = int(
            min(tied)
        )

        selected_row = valid[
            valid["candidate_index"]
            == selected_candidate_index
        ].iloc[0]

        selected_output = str(
            selected_row["candidate_text"]
        )
        selected_utility = float(
            utilities[
                selected_candidate_index
            ]
        )
        selection_status = "mbr_selected"

        for i, ri in enumerate(
            valid.itertuples(index=False)
        ):
            for j, rj in enumerate(
                valid.itertuples(index=False)
            ):
                pair_rows.append({
                    "scenario": SCENARIO_SLUG,
                    "row_id": str(row_id),
                    "candidate_i":
                        int(ri.candidate_index),
                    "candidate_j":
                        int(rj.candidate_index),
                    "is_diagonal":
                        bool(i == j),
                    "cosine_similarity":
                        float(sim[i, j]),
                })

        matrix_rows.append({
            "scenario": SCENARIO_SLUG,
            "row_id": str(row_id),
            "valid_candidate_indices":
                valid_indices,
            "similarity_matrix":
                sim.tolist(),
        })

    elif n == 1:
        only = valid.iloc[0]
        selected_candidate_index = int(
            only["candidate_index"]
        )
        selected_output = str(
            only["candidate_text"]
        )
        selection_status = (
            "single_valid_candidate"
        )
        matrix_rows.append({
            "scenario": SCENARIO_SLUG,
            "row_id": str(row_id),
            "valid_candidate_indices":
                [selected_candidate_index],
            "similarity_matrix": None,
        })

    else:
        selection_status = "inference_failure"
        matrix_rows.append({
            "scenario": SCENARIO_SLUG,
            "row_id": str(row_id),
            "valid_candidate_indices": [],
            "similarity_matrix": None,
        })

    for _, r in group.iterrows():
        idx = int(r["candidate_index"])

        detail = {
            "scenario": SCENARIO_SLUG,
            "row_id": str(row_id),
            "conversation_no":
                r.get("conversation_no", ""),
            "turn_order":
                r.get("turn_order", ""),
            "source_text":
                str(r["source_text"]),
            "candidate_index": idx,
            "candidate_id": f"C{idx}",
            "seed": int(r["seed"]),
            "candidate_text":
                str(r["candidate_text"]),
            "technical_error":
                str(r["technical_error"]),
            "semantic_status":
                str(r["semantic_status"]),
            "semantic_reason":
                str(r["semantic_reason"]),
            "mbr_utility":
                utilities.get(idx, np.nan),
            "is_selected_by_mbr":
                bool(
                    selection_status
                    == "mbr_selected"
                    and idx
                    == int(
                        selected_candidate_index
                    )
                ),
            "s3e_value": s3e_value,
            "valid_candidate_count": int(n),
            "selection_status":
                selection_status,
        }

        for col in [
            "coverage",
            "fragment_count",
            "example_count",
            "max_fragment_len",
            "k_per_fragment",
            "prompt_tokens",
            "completion_tokens",
            "latency_seconds",
            "generation_time_seconds",
        ]:
            if col in r.index:
                detail[col] = r[col]

        detail_rows.append(detail)

    base = group.iloc[0]

    final_row = {
        "scenario": SCENARIO_SLUG,
        "row_id": str(row_id),
        "conversation_no":
            base.get("conversation_no", ""),
        "turn_order":
            base.get("turn_order", ""),
        "source_text":
            str(base["source_text"]),
        "selected_candidate_index":
            selected_candidate_index,
        "selected_candidate":
            ""
            if pd.isna(
                selected_candidate_index
            )
            else f"C{int(selected_candidate_index)}",
        "selected_output": selected_output,
        "selected_utility": selected_utility,
        "s3e_value": s3e_value,
        "valid_candidate_count": int(n),
        "selection_status":
            selection_status,
    }

    for col in [
        "coverage",
        "fragment_count",
        "example_count",
        "max_fragment_len",
        "k_per_fragment",
    ]:
        if col in base.index:
            final_row[col] = base[col]

    selected_rows.append(final_row)


details_df = pd.DataFrame(detail_rows)
selected_df = pd.DataFrame(selected_rows)
pairs_df = pd.DataFrame(pair_rows)

if len(details_df) != 265:
    raise RuntimeError(
        f"Detail rows {len(details_df)} != 265."
    )

if len(selected_df) != 53:
    raise RuntimeError(
        f"Final rows {len(selected_df)} != 53."
    )

print("Candidate detail rows:", len(details_df))
print("Final group rows      :", len(selected_df))
print("\nSelection status:")
print(
    selected_df[
        "selection_status"
    ].value_counts(dropna=False)
)

display(selected_df.head(10))
```
